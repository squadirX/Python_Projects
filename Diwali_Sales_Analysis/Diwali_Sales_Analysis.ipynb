# Exploratory Data Analysis

```python
df.columns
```

```python
# plotting a bar chart -  Number of Buyers by Gender

# Calculate gender counts
gender_counts = df['Gender'].value_counts()

# Replace 'F' with 'Female' and 'M' with 'Male' in the index
gender_counts.index = gender_counts.index.map({'F': 'Female', 'M': 'Male'})

# Function to display both percentage and actual number
def func(pct, allvals):
    absolute = round(pct/100.*sum(allvals), 0)
    return f"{absolute} ({pct:.1f}%)"

# Set the figure size
sns.set(rc={'figure.figsize':(12, 6)})

# Plot the pie chart with both actual count and percentage
plt.pie(gender_counts, 
        labels=gender_counts.index, 
        autopct=lambda pct: func(pct, gender_counts), 
        startangle=90, 
        colors=sns.color_palette('Set2', n_colors=2))

plt.title('Number of Buyers by Gender',fontweight='bold', color='darkblue', fontsize=20)    
plt.axis('equal')
plt.show()
```

```python
# Total Amount( in Value)  vs Age Group
sales_age = df.groupby(['Age Group'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sns.set(rc={'figure.figsize':(20,10)})
ax=sns.barplot(x = 'Age Group',y= 'Amount' ,data = sales_age, hue='Age Group',palette='Set2')
ax.set_ylabel('Value in $')
plt.title('Total Value by Age Group',fontweight='bold', color='darkblue', fontsize=20)
```

```python
# Number of Buyers by Age Group & Gender
ax=sns.countplot(x='Age Group',data=df, hue='Gender')
sns.set(rc={'figure.figsize':(18,10)})
for bars in ax.containers:
    ax.bar_label(bars)
    plt.title('Number of Buyers by Age Group and Gender',fontweight='bold', color='darkblue', fontsize=20)
    ax.set_ylabel('Number of Buyers')
    
```

```python
# plotting a bar chart of Orders for the Top Ten States 

sales_ste = df.groupby(['State'], as_index=False)['Orders'].sum().sort_values(by='Orders', ascending=False).head(10)

sns.set(rc={'figure.figsize':(18,10)})
sns.barplot(x = 'State',y= 'Orders' ,data = sales_ste)
ax = sns.barplot(x='State', y='Orders', data=sales_ste, hue='State', palette='Set2', legend=False)

for bars in ax.containers:
    ax.bar_label(bars)
    plt.title('Total Number of Orders by Top Ten States',fontweight='bold', color='darkblue', fontsize=20)
```

```python
# plotting a bar chart of Amount for the Top Ten States 

sales_ste = df.groupby(['State'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False).head(10)

sns.set(rc={'figure.figsize':(18,10)})
sns.barplot(x = 'State',y= 'Amount' ,data = sales_ste, hue='State', palette='Set2')
plt.title('Total Value Contribution by Top Ten States',fontweight='bold', color='darkblue', fontsize=20)
```

```python
# plotting a bar chart of Amount by Marital Status & Gender

sales_ms = df.groupby(['Marital_Status', 'Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sns.set(rc={'figure.figsize':(20,10)})
sns.barplot(x = 'Marital_Status',y= 'Amount' ,data = sales_ms, hue='Gender')
plt.title('Value Contribution by Marital Status and Gender',fontweight='bold', color='darkblue', fontsize=20)
```

```python
Cat_data = df['Product_Category'].value_counts().nlargest(10).index      
ax = sns.countplot(x='Product_Category', data=df[df['Product_Category'].isin(Cat_data)], hue='Gender')
sns.set(rc={'figure.figsize':(20,5)})
for bars in ax.containers:
    ax.bar_label(bars)
    plt.title('Number of user Top 10 Product Categories by Gender Distribution',fontweight='bold', color='darkblue', fontsize=20 )
```

```python
 # plotting a bar chart of Amount by Product Category & Gender

sales_ms = df.groupby(['Product_Category', 'Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False).head(18)
sns.set(rc={'figure.figsize':(20,7)})
sns.barplot(x = 'Product_Category',y= 'Amount' ,data = sales_ms, hue='Gender')
plt.title('Top 10 Product Categories by Sales Amount with Gender Comparison',fontweight='bold', color='darkblue', fontsize=20)

```

```python
Cat_data = df['Product_ID'].value_counts().nlargest(10).index      
ax = sns.countplot(x='Product_ID', data=df[df['Product_ID'].isin(Cat_data)], hue='Gender')
sns.set(rc={'figure.figsize':(20,5)})
for bars in ax.containers:
    ax.bar_label(bars)
    plt.title('Top 10 Products by Sales Amount (with Gender Comparison)',fontweight='bold', color='darkblue', fontsize=20)
```

```python
 # plotting a bar chart of Amount by Product ID & Gender

sales_ms = df.groupby(['Product_ID', 'Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False).head(10)
sns.set(rc={'figure.figsize':(20,7)})
sns.barplot(x = 'Product_ID',y= 'Amount' ,data = sales_ms, hue='Gender', palette='Set2')
plt.title('Top 10 Products by Sales Value with Gender Comparison',fontweight='bold', color='darkblue', fontsize=20)
```

Summary Findings:
The analysis reveals several key insights: Female buyers outnumber male buyers, leading to higher total spending by women. Among age groups, individuals aged 26-35, predominantly female, are the highest spenders. Regionally, Uttar Pradesh leads in both order volume and spending, followed by Maharashtra and Karnataka. Unmarried women are the top spenders, surpassing both married women and male buyers in total expenditure. Professionally, the IT sector tops both participation and spending, followed by Healthcare and Aviation. When examining product categories, Clothing ranks highest by quantity sold, followed by Food and Footwear, while Food generates the most revenue, with Footwear and Clothing in second and third places, respectively.
